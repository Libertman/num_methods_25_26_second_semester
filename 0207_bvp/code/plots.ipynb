```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go
```

```python
solution_df = pd.read_csv("../data/solution.csv")
error_df = pd.read_csv("../data/error.csv")
accuracy_df = pd.read_csv("../data/accuracy.csv")
step_size_df = pd.read_csv("../data/step_size.csv")

sol_h1 = solution_df[solution_df["type"] == "h1"]
sol_h2 = solution_df[solution_df["type"] == "h2"]
err_h1 = error_df[error_df["type"] == "h1"]
err_h2 = error_df[error_df["type"] == "h2"]
```

```python
# 1. График точного и численных решений (h1 = 0.1, h2 = 0.05)
fig_solution = go.Figure()

fig_solution.add_trace(
    go.Scatter(
        x=sol_h2["x"],
        y=sol_h2["y_exact"],
        mode="lines",
        name="Точное решение y = e^x",
        line=dict(color="#1f77b4", width=3)
    )
)

fig_solution.add_trace(
    go.Scatter(
        x=sol_h1["x"],
        y=sol_h1["y_numeric"],
        mode="lines+markers",
        name="Численное решение (h = 0.1)",
        marker=dict(size=8, symbol="circle", color="#d62728"),
        line=dict(dash="dash", width=1.5, color="#d62728")
    )
)

fig_solution.add_trace(
    go.Scatter(
        x=sol_h2["x"],
        y=sol_h2["y_numeric"],
        mode="lines+markers",
        name="Численное решение (h = 0.05)",
        marker=dict(size=6, symbol="diamond", color="#2ca02c"),
        line=dict(dash="dot", width=1.5, color="#2ca02c")
    )
)

fig_solution.update_layout(
    title="Сравнение точного и численных решений методом факторизации",
    xaxis_title="x",
    yaxis_title="y(x)",
    template="ggplot2",
    hovermode="x unified",
    height=650,
    width=1200
)

fig_solution.show()
```

```python
# 2. График ошибки на отрезке для h1 и h2
fig_error = go.Figure()

fig_error.add_trace(
    go.Scatter(
        x=err_h1["x"],
        y=err_h1["error"],
        mode="lines+markers",
        name="Погрешность (h = 0.1)",
        marker=dict(size=8, symbol="circle", color="#d62728"),
        line=dict(width=2, color="#d62728")
    )
)

fig_error.add_trace(
    go.Scatter(
        x=err_h2["x"],
        y=err_h2["error"],
        mode="lines+markers",
        name="Погрешность (h = 0.05)",
        marker=dict(size=6, symbol="diamond", color="#2ca02c"),
        line=dict(width=2, color="#2ca02c")
    )
)

fig_error.update_layout(
    title="График абсолютной погрешности на отрезке [0, 1]",
    xaxis_title="x",
    yaxis_title="|y_num - y_exact|",
    template="ggplot2",
    hovermode="x unified",
    height=650,
    width=1200
)

fig_error.show()
```

```python
# 3. График точности (фактическая погрешность от eps)
fig_accuracy = go.Figure()

fig_accuracy.add_trace(
    go.Scatter(
        x=accuracy_df["eps"],
        y=accuracy_df["max_error"],
        mode="lines+markers",
        name="Фактическая погрешность",
        marker=dict(size=9, symbol="square", color="#9467bd"),
        line=dict(width=2.5, color="#9467bd")
    )
)

fig_accuracy.add_trace(
    go.Scatter(
        x=accuracy_df["eps"],
        y=accuracy_df["eps"],
        mode="lines",
        name="Опорная прямая y = eps",
        line=dict(dash="dash", color="#7f7f7f", width=1.5)
    )
)

fig_accuracy.update_layout(
    title="Зависимость фактической ошибки от заданной точности",
    xaxis_title="Заданная точность eps",
    yaxis_title="Фактическая погрешность",
    template="ggplot2",
    hovermode="x unified",
    height=650,
    width=1200
)

fig_accuracy.update_xaxes(type="log", tickformat=".1e")
fig_accuracy.update_yaxes(type="log", tickformat=".1e")

fig_accuracy.show()
```

```python
# 4. График изменения адаптивного шага по отрезку при eps = 1e-6
fig_step = go.Figure()

fig_step.add_trace(
    go.Scatter(
        x=step_size_df["x"],
        y=step_size_df["h"],
        mode="lines+markers",
        name="Шаг интегрирования h(x)",
        marker=dict(size=8, symbol="cross", color="#ff7f0e"),
        line=dict(width=2, color="#ff7f0e", shape="hv")
    )
)

fig_step.update_layout(
    title="Изменение адаптивного шага по отрезку при eps = 1e-6",
    xaxis_title="x",
    yaxis_title="Шаг h",
    template="ggplot2",
    hovermode="x unified",
    height=650,
    width=1200
)

fig_step.show()
```
