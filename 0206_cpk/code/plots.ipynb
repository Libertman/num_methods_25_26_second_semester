```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python
solution = pd.read_csv("../data/solution.csv")
error = pd.read_csv("../data/error.csv")
accuracy = pd.read_csv("../data/accuracy.csv")
```

```python
fig_solution = go.Figure()

fig_solution.add_trace(
    go.Scatter(
        x=solution["x"],
        y=solution["y_numeric"],
        mode="lines+markers",
        name="Численное решение"
    )
)

fig_solution.add_trace(
    go.Scatter(
        x=solution["x"],
        y=error["y_exact"],
        mode="lines",
        name="Точное решение"
    )
)

fig_solution.update_layout(
    title="Сравнение точного и численного решений",
    xaxis_title="x",
    yaxis_title="y",
    template="ggplot2",
    hovermode="x unified",
    height=650,
    width=1200
)

fig_solution.show()
```

```python
fig_error = go.Figure()

fig_error.add_trace(
    go.Scatter(
        x=solution["x"],
        y=error["error"],
        mode="lines+markers",
        name="Абсолютная ошибка"
    )
)

fig_error.update_layout(
    title="Ошибка численного решения",
    xaxis_title="x",
    yaxis_title="|y_num - y_exact|",
    template="ggplot2",
    hovermode="x unified",
    height=650,
    width=1200
)

fig_error.show()
```

```python
fig_accuracy = go.Figure()

fig_accuracy.add_trace(
    go.Scatter(
        x=accuracy["eps"],
        y=accuracy["max_error"],
        mode="lines+markers",
        name="Фактическая ошибка"
    )
)

fig_accuracy.update_layout(
    title="Зависимость фактической ошибки от заданной точности",
    xaxis_title="Заданная точность",
    yaxis_title="Максимальная фактическая ошибка",
    template="ggplot2",
    hovermode="x unified",
    height=650,
    width=1200
)

fig_accuracy.update_yaxes(
    type="log",
    tickformat=".1e",
)

fig_accuracy.update_xaxes(
    type="log",
        tickformat=".1e"
    )

fig_accuracy.show()
```
