## Сравнение решения интегралов изученными методами
* Метод 1: Метод трапеций (Ньютон–Котес, Лаб. №3)
* Метод 2: Метод Гаусса–Лобатто (Лаб. №4)

Все графики построены исключительно с использованием библиотеки **Plotly**.

```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go
import plotly.io as pio

# Загрузка данных численного эксперимента
df = pd.read_csv("../data/comparison_data.csv")
trap_df = df[df["method"] == "Trapezoidal"].copy()
lob_df = df[df["method"] == "Gauss-Lobatto"].copy()

print("Загружено строк:", len(df))
display(df.head() if "display" in globals() else df.head())
```

```python
# ---------------------------------------------------------------------
# График 1: Вычислительные затраты (N_calls) от заданной точности eps
# ---------------------------------------------------------------------
fig1 = go.Figure()

fig1.add_trace(go.Scatter(
    x=trap_df["eps"],
    y=trap_df["func_calls"],
    mode="lines+markers",
    name="Метод трапеций (Ньютон–Котес)",
    line=dict(color="#EF553B", width=2.5),
    marker=dict(size=8, symbol="square")
))

fig1.add_trace(go.Scatter(
    x=lob_df["eps"],
    y=lob_df["func_calls"],
    mode="lines+markers",
    name="Метод Гаусса–Лобатто (4-5 узлов)",
    line=dict(color="#00CC96", width=2.5),
    marker=dict(size=8, symbol="circle")
))

fig1.update_layout(
    title=dict(
        text="Зависимость числа вызовов функции от заданной точности eps",
        font=dict(size=18, family="Arial")
    ),
    xaxis=dict(
        title="Заданная точность eps",
        type="log",
        autorange="reversed",
        showgrid=True,
        gridcolor="#E5E5E5",
        tickformat=".1e"
    ),
    yaxis=dict(
        title="Число вызовов функции N_calls",
        type="log",
        showgrid=True,
        gridcolor="#E5E5E5",
        tickformat=".1e"
    ),
    template="plotly_white",
    legend=dict(x=0.03, y=0.97, bgcolor="rgba(255,255,255,0.85)", bordercolor="#CCCCCC", borderwidth=1),
    width=1000,
    height=600
)

fig1.write_image("../data/calls_vs_eps.png", scale=2)
fig1.show()
```

```python
# ---------------------------------------------------------------------
# График 2: Эффективность методов (погрешность от числа вызовов)
# ---------------------------------------------------------------------
fig2 = go.Figure()

fig2.add_trace(go.Scatter(
    x=trap_df["func_calls"],
    y=trap_df["error"],
    mode="lines+markers",
    name="Метод трапеций (Ньютон–Котес)",
    line=dict(color="#EF553B", width=2.5),
    marker=dict(size=8, symbol="square")
))

fig2.add_trace(go.Scatter(
    x=lob_df["func_calls"],
    y=lob_df["error"],
    mode="lines+markers",
    name="Метод Гаусса–Лобатто (4-5 узлов)",
    line=dict(color="#00CC96", width=2.5),
    marker=dict(size=8, symbol="circle")
))

fig2.update_layout(
    title=dict(
        text="Зависимость фактической ошибки от вычислительных затрат",
        font=dict(size=18, family="Arial")
    ),
    xaxis=dict(
        title="Число вызовов функции N_calls",
        type="log",
        showgrid=True,
        gridcolor="#E5E5E5",
        tickformat=".1e"
    ),
    yaxis=dict(
        title="Фактическая погрешность |I_num - I_exact|",
        type="log",
        showgrid=True,
        gridcolor="#E5E5E5",
        tickformat=".1e"
    ),
    template="plotly_white",
    legend=dict(x=0.6, y=0.97, bgcolor="rgba(255,255,255,0.85)", bordercolor="#CCCCCC", borderwidth=1),
    width=1000,
    height=600
)

fig2.write_image("../data/error_vs_calls.png", scale=2)
fig2.show()
```

```python
# ---------------------------------------------------------------------
# График 3: Соответствие фактической погрешности заданной точности
# ---------------------------------------------------------------------
fig3 = go.Figure()

fig3.add_trace(go.Scatter(
    x=trap_df["eps"],
    y=trap_df["error"],
    mode="lines+markers",
    name="Метод трапеций",
    line=dict(color="#EF553B", width=2.5)
))

fig3.add_trace(go.Scatter(
    x=lob_df["eps"],
    y=lob_df["error"],
    mode="lines+markers",
    name="Метод Гаусса–Лобатто",
    line=dict(color="#00CC96", width=2.5)
))

fig3.add_trace(go.Scatter(
    x=df["eps"],
    y=df["eps"],
    mode="lines",
    name="Биссектриса y = x (допуск)",
    line=dict(color="#666666", width=1.5, dash="dash")
))

fig3.update_layout(
    title=dict(
        text="Соответствие фактической погрешности заданной точности (проверка биссектрисы)",
        font=dict(size=18, family="Arial")
    ),
    xaxis=dict(
        title="Заданная точность eps",
        type="log",
        autorange="reversed",
        showgrid=True,
        gridcolor="#E5E5E5",
        tickformat=".1e"
    ),
    yaxis=dict(
        title="Фактическая погрешность",
        type="log",
        showgrid=True,
        gridcolor="#E5E5E5",
        tickformat=".1e"
    ),
    template="plotly_white",
    legend=dict(x=0.03, y=0.97, bgcolor="rgba(255,255,255,0.85)", bordercolor="#CCCCCC", borderwidth=1),
    width=1000,
    height=600
)

fig3.write_image("../data/error_vs_eps.png", scale=2)
fig3.show()
```
