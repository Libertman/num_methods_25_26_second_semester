```python
import numpy as np
import plotly.graph_objects as go
```

```python
data = np.loadtxt("../../data/results.txt")

eps = data[:,0]
I = data[:,1]
I_rich = data[:,2]
h = data[:,3]
```

```python
I_exact = 348.25 * np.sin(20) + 35 * np.cos(20)
error = abs(I - I_exact)
error_rich = abs(I_rich - I_exact)
```

```python
I_exact
```

```python
def P(x):
	return x ** 5 - 6.2 * x ** 3 + 3.5 * x ** 2 - 7 * x



def func(x):
	return P(x) * np.cos(2 * x)
```

```python
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=np.linspace(-10, 10, 10000), y=func(np.linspace(-10, 10, 10000))
))

fig.update_layout(
    hovermode="x unified",
    template="ggplot2",
    title="Функция (от -10 до 10)",
    xaxis_title="x",
    yaxis_title="y",
    height=650,
    width=1200
)

fig.update_yaxes(
    tickformat=".1e"
)

fig.update_xaxes(
    tickformat=".1e"
)
```

```python
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=np.linspace(-4, 4, 10000), y=func(np.linspace(-4, 4, 10000))
))

fig.update_layout(
    hovermode="x unified",
    template="ggplot2",
    title="Функция (от -4 до 4)",
    xaxis_title="x",
    yaxis_title="y",
    height=650,
    width=1200
)

fig.update_yaxes(
    tickformat=".1e"
)

fig.update_xaxes(
    tickformat=".1e"
)
```

```python
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=eps, y=error,
    mode='lines+markers',
    name='Error'
))

fig.add_trace(go.Scatter(
    x=eps, y=error_rich,
    mode='lines+markers',
    name='Error (Richardson)'
))

# биссектриса
fig.add_trace(go.Scatter(
    x=eps, y=eps,
    mode='lines',
    name='y=x',
    line=dict(dash='dash')
))

fig.update_layout(
    xaxis_type="log",
    hovermode="x unified",
    template="ggplot2",
    yaxis_type="log",
    title="Ошибка vs точность",
    xaxis_title="eps",
    yaxis_title="error",
    height=650,
    width=1200
)

fig.update_yaxes(
    tickformat=".1e"
)

fig.update_xaxes(
    tickformat=".1e"
)

fig.show()
```

```python
fig = go.Figure()

h_log2 = np.log2(h)
error_log2 = np.log2(error)
error_rich_log2 = np.log2(error_rich)

fig.add_trace(go.Scatter(
    x=h_log2,
    y=error_log2,
    mode='lines+markers',
    name='Error'
))

fig.add_trace(go.Scatter(
    x=h_log2,
    y=error_rich_log2,
    mode='lines+markers',
    name='Error (Richardson)'
))

fig.update_layout(
    title="log2(error) vs log2(h)",
    xaxis_title="log2(h)",
    yaxis_title="log2(error)",
    height=650,
    width=1200
)


fig.show()
```

```python
coef = np.polyfit(np.log2(h), np.log2(error), 1)
p = coef[0]
C = 2**coef[1]
```
